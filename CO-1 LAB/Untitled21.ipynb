## 14. Question Answering using a Pre-trained Transformer

```python
from transformers import AutoModelForQuestionAnswering, AutoTokenizer
import torch

model_name = 'distilbert-base-cased-distilled-squad'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

context_text = "Google Colab is a hosted Jupyter notebook service that requires no setup to use."
question_text = "What is Google Colab?"

inputs = tokenizer(question_text, context_text, return_tensors="pt")
with torch.no_grad():
    outputs = model(**inputs)

answer_start = torch.argmax(outputs.start_logits)
answer_end = torch.argmax(outputs.end_logits) + 1
answer = tokenizer.convert_tokens_to_string(tokenizer.convert_ids_to_tokens(inputs["input_ids"][0][answer_start:answer_end]))

print({'answer': answer})
```

## 17. Generate contextual embeddings for a sentence using BERT

```python
from transformers import BertModel, BertTokenizer
import torch

bert_tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
bert_model = BertModel.from_pretrained('bert-base-uncased')

sentence_emb = "Hugging Face transformers provide great APIs."
inputs_emb = bert_tokenizer(sentence_emb, return_tensors="pt")
with torch.no_grad():
    outputs_emb = bert_model(**inputs_emb)

# Mean pooling over token embeddings to represent the sentence
mean_embedding = torch.mean(outputs_emb.last_hidden_state, dim=1)
print("Contextual Sentence Embedding Shape:", mean_embedding.shape)
```

## 18. Compare contextual embeddings generated for two semantically similar sentences

```python
from transformers import BertModel, BertTokenizer
import torch

bert_tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
bert_model = BertModel.from_pretrained('bert-base-uncased')

sentence1 = "The weather is warm today."
sentence2 = "It is quite hot outside today."

def get_embedding(text):
    inputs = bert_tokenizer(text, return_tensors="pt")
    with torch.no_grad():
        outputs = bert_model(**inputs)
    return torch.mean(outputs.last_hidden_state, dim=1)

emb1 = get_embedding(sentence1)
emb2 = get_embedding(sentence2)

cos_sim = torch.nn.functional.cosine_similarity(emb1, emb2)
print(f"Cosine Similarity between sentences: {cos_sim.item():.4f}")
```
